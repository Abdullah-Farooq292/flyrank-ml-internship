## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (like "flag pages that are stale AND still getting traffic") only combines a couple of signals in a hand-chosen way. A trained model can weigh many signals together and find combinations a human wouldn't think to check. This was shown directly in Notebook 02: the hand-written rule scored 0.680 Precision@50, while a simple depth-2 tree still found different, useful patterns despite having only 3 questions to work with — and the starter pipeline's fuller models (logistic regression, random forest) improved Precision@50 up to 0.740 versus the baseline rule's 0.240. This shows ML can meaningfully outperform a fixed rule once given more signals to combine.
